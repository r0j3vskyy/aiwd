# Analiza danych i wizualizacje

Notebook korzysta z walidowanych funkcji z `src/analysis_project` i zapisuje wykresy w `figures`.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt

ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'src' / 'analysis_project').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

from analysis_project import (
    births_by_year_and_sex,
    load_names,
)

RAW = ROOT / 'data' / 'raw'
FIGURES = ROOT / 'figures'
FIGURES.mkdir(parents=True, exist_ok=True)
```

```python
names = load_names(RAW / 'imiona.xlsx')
births = births_by_year_and_sex(names)
births.plot(figsize=(10, 5), title='Liczba urodzeń według roku i płci')
plt.xlabel('Rok')
plt.ylabel('Liczba')
plt.tight_layout()
plt.savefig(FIGURES / 'births_by_year_and_sex.png', dpi=150)
plt.show()
```
